```python
# ============================================================
# Project 5: GoldCare Connect Analytics Dashboard (Capstone)
# Analyst: Adebowale Adebare
# Purpose: Simulated national analytics dashboard showing
#          the value of GoldCare Connect platform adoption
# Data Sources: CQC (Project 1), NHS Workforce (Project 2),
#               Skills for Care (Project 3)
# Date: September 2026
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
# Combined dataset pulling real figures from Projects 1, 2 and 3
# This simulates what the GoldCare Connect platform would track nationally

data = {
    'Region': [
        'North East', 'North West', 'Yorkshire and the Humber',
        'East Midlands', 'West Midlands', 'East of England',
        'London', 'South East', 'South West'
    ],
    # From Project 1 — CQC Analysis
    'Total_Care_Homes': [689, 1713, 1340, 1331, 1514, 1467, 1184, 2579, 1740],
    'Outstanding_Pct': [6.0, 3.4, 4.9, 6.5, 2.7, 5.4, 2.6, 4.5, 5.6],
    'Good_Pct': [74.0, 72.0, 71.0, 69.0, 71.0, 72.0, 73.0, 71.0, 72.0],
    'Requires_Improvement_Pct': [16.5, 18.5, 18.0, 19.0, 21.0, 17.5, 22.0, 19.0, 17.5],
    'Inadequate_Pct': [0.5, 0.7, 0.5, 0.5, 0.7, 0.4, 1.2, 0.7, 0.5],
    # From Project 2 — NHS Workforce
    'NHS_Headcount_2026': [86746, 236825, 157828, 126469, 167459, 139457, 259976, 211591, 151694],
    'NHS_Leavers_Annual': [11242, 27763, 21917, 21773, 24140, 23918, 26250, 36555, 25266],
    # From Project 3 — Skills for Care
    'Care_Workers': [41000, 105000, 78000, 79000, 93000, 90000, 114000, 131000, 88000],
    'Turnover_Rate_Pct': [27.1, 26.0, 27.7, 27.5, 25.9, 26.1, 22.1, 27.2, 27.9],
    'Vacancy_Rate_Pct': [5.6, 7.3, 6.6, 8.0, 7.7, 7.9, 9.5, 7.7, 7.3],
    'Mean_Hourly_Rate': [12.16, 12.22, 12.19, 12.05, 12.05, 12.24, 12.52, 12.33, 12.37]
}

df = pd.DataFrame(data)

# Calculate derived columns
df['Annual_Leavers'] = (df['Care_Workers'] * df['Turnover_Rate_Pct'] / 100).astype(int)
df['Vacant_Posts'] = (df['Care_Workers'] * df['Vacancy_Rate_Pct'] / 100).astype(int)

# Agency costs (30% premium, 8 shifts per leaver, 8 hour shifts)
df['Agency_Cost_Per_Home'] = 10144  # From Project 3
df['Platform_Cost_Per_Home'] = 2588  # From Project 3
df['Saving_Per_Home'] = 7556       # From Project 3

# Regional total costs
df['Total_Agency_Cost'] = df['Total_Care_Homes'] * df['Agency_Cost_Per_Home']
df['Total_Platform_Cost'] = df['Total_Care_Homes'] * df['Platform_Cost_Per_Home']
df['Total_Saving'] = df['Total_Care_Homes'] * df['Saving_Per_Home']

print(f"Dataset created: {df.shape}")
print(f"\nColumns: {df.columns.tolist()}")
df[['Region', 'Total_Care_Homes', 'Annual_Leavers', 'Total_Saving']].head()
```
